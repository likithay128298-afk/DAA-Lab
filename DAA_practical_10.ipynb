```python
def find(parent, i):
    if parent[i] != i:
        parent[i] = find(parent, parent[i])
    return parent[i]


def union(parent, rank, x, y):
    x = find(parent, x)
    y = find(parent, y)

    if x == y:
        return False

    if rank[x] < rank[y]:
        parent[x] = y
    elif rank[x] > rank[y]:
        parent[y] = x
    else:
        parent[y] = x
        rank[x] += 1

    return True


V = int(input("Enter number of vertices: "))
E = int(input("Enter number of edges: "))

edges = []

print("Enter edges (u v weight), using vertices 0 to", V - 1)
for _ in range(E):
    u, v, w = map(int, input().split())
    edges.append((w, u, v))

edges.sort()

parent = list(range(V))
rank = [0] * V
total_cost = 0
count = 0

print("\nEdges in Minimum Spanning Tree:")

for w, u, v in edges:
    if union(parent, rank, u, v):
        print(u, "-", v, ":", w)
        total_cost += w
        count += 1

        if count == V - 1:
            break

if count == V - 1:
    print("Minimum cost =", total_cost)
else:
    print("Graph is disconnected!")
```
